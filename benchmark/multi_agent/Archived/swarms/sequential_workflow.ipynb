```python
import os

# Must set no_proxy BEFORE importing libraries that cache proxy settings
os.environ["no_proxy"] = os.environ.get("no_proxy", "") + ",gpu14"
os.environ["NO_PROXY"] = os.environ.get("NO_PROXY", "") + ",gpu14"

import litellm
from swarms import Agent, SequentialWorkflow

# Register custom model so litellm can resolve max_tokens
litellm.model_cost["openai/Qwen/Qwen3-4B-Instruct-2507"] = {
    "max_tokens": 32768,
    "input_cost_per_token": 0,
    "output_cost_per_token": 0,
}

LLM_BASE_URL = "http://gpu14:8000/v1"
LLM_API_KEY = os.getenv("OPENROUTER_API_KEY")

common_llm_args = {
    "base_url": LLM_BASE_URL,
    "api_key": LLM_API_KEY,
}

# Create specialized agents for a content creation pipeline
researcher = Agent(
    agent_name="Researcher",
    system_prompt="You are an expert researcher. Gather comprehensive information on the given topic.",
    model_name="openai/Qwen/Qwen3-4B-Instruct-2507",
    llm_args={
        **common_llm_args,
        "user": "Researcher",
        "headers": {"x-sglang-agent-id": "Researcher"},
    },
    top_p=None,
    max_loops=1,
    dynamic_temperature_enabled=True,
)

analyzer = Agent(
    agent_name="Analyzer",
    system_prompt="You are a data analyst. Identify key trends, patterns, and insights from research data.",
    model_name="openai/Qwen/Qwen3-4B-Instruct-2507",
    llm_args={
        **common_llm_args,
        "user": "Analyzer",
        "headers": {"x-sglang-agent-id": "Analyzer"},
    },
    top_p=None,
    max_loops=1,
    dynamic_temperature_enabled=True,
)

writer = Agent(
    agent_name="Writer",
    system_prompt="You are a professional writer. Create engaging content based on research and analysis.",
    model_name="openai/Qwen/Qwen3-4B-Instruct-2507",
    llm_args={
        **common_llm_args,
        "user": "Writer",
        "headers": {"x-sglang-agent-id": "Writer"},
    },
    top_p=None,
    max_loops=1,
    dynamic_temperature_enabled=True,
)

# Create the sequential workflow
workflow = SequentialWorkflow(agents=[researcher, analyzer, writer])

# Execute the workflow
result = workflow.run("The impact of artificial intelligence on healthcare")
print(result)
```
